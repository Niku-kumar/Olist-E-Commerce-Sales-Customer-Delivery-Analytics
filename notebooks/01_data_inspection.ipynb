```python
sales_data = (
    order_items_clean
    .merge(orders_clean, on="order_id", how="left")
    .merge(customers_clean, on="customer_id", how="left")
    .merge(products_clean, on="product_id", how="left")
    .merge(category_translation_clean, on="product_category_name", how="left")
    .merge(sellers_clean, on="seller_id", how="left")
)

print("Sales dataset shape:", sales_data.shape)
```

```python
sales_data["item_revenue"] = sales_data["price"]

sales_data["total_item_value"] = (
    sales_data["price"] + sales_data["freight_value"]
)

print(
    sales_data[
        ["price", "freight_value", "item_revenue", "total_item_value"]
    ].head()
)
```

```python
print("Missing values after merging:")
print(sales_data.isnull().sum().sort_values(ascending=False).head(10))
```

```python
sales_data["product_category_name_english"] = (
    sales_data["product_category_name_english"]
    .fillna("Unknown")
)

print(
    sales_data["product_category_name_english"]
    .isnull()
    .sum()
)
```

```python
print("Final sales dataset shape:", sales_data.shape)
print("Unique orders:", sales_data["order_id"].nunique())
print("Unique products:", sales_data["product_id"].nunique())
print("Unique customers:", sales_data["customer_id"].nunique())
```

```python
total_revenue = sales_data["item_revenue"].sum()
total_orders = sales_data["order_id"].nunique()
total_customers = sales_data["customer_id"].nunique()
total_products = sales_data["product_id"].nunique()

average_order_value = total_revenue / total_orders

print("Total Revenue:", round(total_revenue, 2))
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Products Sold:", total_products)
print("Average Order Value:", round(average_order_value, 2))
```

```python
total_unique_customers = sales_data["customer_unique_id"].nunique()

print("Actual Unique Customers:", total_unique_customers)
```

```python
monthly_sales = (
    sales_data
    .groupby(["order_year", "order_month"], as_index=False)["item_revenue"]
    .sum()
    .sort_values(["order_year", "order_month"])
)

monthly_sales["year_month"] = (
    monthly_sales["order_year"].astype(str)
    + "-"
    + monthly_sales["order_month"].astype(str).str.zfill(2)
)

print(monthly_sales)
```

```python
sales_data.to_csv(
    "../data/cleaned/sales_data.csv",
    index=False
)

print("Final integrated sales dataset saved successfully!")
```
